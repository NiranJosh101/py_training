# Day 1: Division and Square-root
Write a function called divide_or_square that takes one argument (a 
number), and returns the square root of the number if it is divisible by 
5, returns its remainder if it is not divisible by 5. For example, if you 
pass 10 as an argument, then your function should return 3.16 as the 
square root

```python
def divide_or_square(num):
    if num%5==0:
        return num**0.5
    else:
        remainder=num%5
        return remainder
    

divide_or_square(10)
```

#### 1. Given an integer array num return true if any value appears at least twice in the array and false if every element is distict

```python
def check_dup(num):
    lst = []

    for n in num:
        if n in lst:
            return True
        lst.append(n)
    return False
```

```python
num = [1, 2, 1, 4]


check_dup(num)
```

```python
# optimized aproach
# using set() bc i s stored as an hash table That's why membership checks are approximately O(1).The lookup time stays roughly constant even as the set grows.

# Membership testing in a list is O(n). The work grows linearly with the size of the list.

def check_dup(num):
    lst = set()

    for n in num:
        if n in lst:
            return True
        lst.add(n)
    return False
```

```python
num = [1, 2, 1, 4]
check_dup(num)
```

#### 3. Find all numbers disappeard in an array

given a array nums of n integers where num[i] is in range [1, n], return an array of all the integer in the range[1, n] that do not appear in nums

```python


def find_disapp(nums):
    miss = []

    # remove dups
    num = set(nums)

    # loop through the range
    for i in range(1, len(nums)+1):
        if i not in num:
            miss.append(i)

    return miss 
```

```python
nums = [4,3,2,7,8,2,3,1]

find_disapp(nums)
```
